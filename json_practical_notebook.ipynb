# JSON in Python — useful functionalities
This notebook demonstrates **all JSON functionalities to be used in the project**, plus a few closely related ones that are commonly useful.

You can run every cell.

## 1. Importing modules

This cell imports the core modules used throughout the notebook.

json allows conversion between Python objects and JSON format.

os is used to interact with the file system, such as checking if a file exists.

datetime helps store timestamps in a JSON-safe string format.
These imports form the foundation for working with persistent data.

```python
import json
import os
from datetime import datetime
```

## 2. Python object → JSON string (`json.dumps`)

json.dumps() converts a Python object into a JSON-formatted string.

The data is not written to a file; it exists only as a string in memory.

This is useful for debugging, logging, or sending data over a network.

The output is plain text that follows JSON rules.

Think of this as serialization without storage.

```python
data = {
    "project": "remind",
    "number of weeks": 4,
    "skills learned": ["Python", "agentic-ai", "streamlit"]
}

json_string = json.dumps(data)
json_string
```

## 3. JSON string → Python object (`json.loads`)

json.loads() converts a JSON string back into Python objects.

It reverses the effect of json.dumps().

After loading, the data behaves like normal Python dictionaries or lists.

This is commonly used when receiving JSON from APIs or files read as text.

The result lives entirely in memory.

```python
python_obj = json.loads(json_string)
python_obj
```

## 4. Writing Python data to a JSON file (`json.dump`)

json.dump() is used to save Python data (like dictionaries or lists) into a file in JSON format.

It converts in-memory Python objects into readable text and writes them to disk.

This allows data to persist even after the program stops running.

The function requires an open file object and overwrites the file when opened in write mode.

It is commonly used for local storage of app data and state.

```python
file_name = "example.json"

with open(file_name, "w") as f:
    json.dump(data, f, indent=4)

os.path.exists(file_name)
```

## 5. Reading JSON data from a file (`json.load`)

json.load() reads JSON data from a file and converts it into Python objects.

The file must be opened in read mode.

After loading, the data can be accessed like any Python dictionary or list.

This function restores data that was previously saved using json.dump().

It is the primary way apps recover saved state.

```python
with open(file_name, "r") as f:
    loaded_data = json.load(f)

loaded_data
```

## 6. Safe loading pattern 

This pattern prevents errors when a JSON file does not yet exist.

If the file is missing, a default Python structure is returned instead.

This ensures the rest of the program can run safely without crashing.

It is a defensive programming technique used in real-world applications.

```python
def load_data_safe(filename):
    if not os.path.exists(filename):
        return {"items": []}
    with open(filename, "r") as f:
        return json.load(f)

load_data_safe("non_existent.json")
```

## 7. Updating part of JSON data and saving again

This cell demonstrates the load → modify → save pattern.

Existing data is first loaded from the file into Python.

A new field is added using a timestamp converted to a string.

The updated data is then written back to the same file.

This is how persistent state is safely updated.

```python
data = load_data_safe(file_name)
data["last_updated"] = datetime.now().isoformat()

with open(file_name, "w") as f:
    json.dump(data, f, indent=4)

data
```

## 8. Appending to JSON-stored lists (very common pattern)

This pattern is used when data grows over time, such as logs or histories.

The JSON file is loaded, modified, and then saved again.

setdefault() ensures the list exists before appending to it.

Each function call updates the file without losing previous entries.

This is extremely common in apps that track user activity.

```python
def append_item(filename, item):
    data = load_data_safe(filename)
    data.setdefault("items", []).append(item)
    with open(filename, "w") as f:
        json.dump(data, f, indent=4)

append_item("items.json", {"value": 1})
append_item("items.json", {"value": 2})

load_data_safe("items.json")
```

## 9. Common JSON-compatible Python types
- dict → object
- list → array
- str → string
- int/float → number
- True/False → true/false
- None → null

## 10. Key takeaways
- `dump / load` → file-based persistence
- `dumps / loads` → string-based conversion
- JSON is ideal for small-to-medium local storage
- Always guard against missing files
